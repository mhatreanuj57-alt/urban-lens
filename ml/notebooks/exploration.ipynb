# UrbanLens Data Exploration

Initial exploration of the UrbanLens dataset.

```python
import pandas as pd
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
```

```python
# Load annotations
data_dir = Path('../data/urbanlens')
image_files = list((data_dir / 'images' / 'train').glob('*'))
label_files = list((data_dir / 'labels' / 'train').glob('*'))

print(f'Training images: {len(image_files)}')
print(f'Training labels: {len(label_files)}')
```

```python
# Class distribution
class_names = ['pothole', 'garbage', 'damaged_streetlight', 'waterlogging', 'illegal_dumping']
class_counts = {name: 0 for name in class_names}

for label_file in label_files:
    with open(label_file) as f:
        for line in f:
            class_id = int(line.split()[0])
            class_counts[class_names[class_id]] += 1

plt.bar(class_counts.keys(), class_counts.values())
plt.title('Class Distribution')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
